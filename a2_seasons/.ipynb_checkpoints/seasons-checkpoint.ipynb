# Seasons and hunger: lean-season food prices and the Cadre Harmonisé

**Research question 1.** How strong is the lean-season rise in staple food prices, how does it differ by food and region,
and does it coincide with the lean-season peak in the share of people in Cadre Harmonisé phase 3 or worse?

- **RQ1a** (Chart A, heatmap): in which months and for which foods are prices highest relative to the rest of the year,
  and do they match the season when the Cadre Harmonisé counts the most people in crisis?
- **RQ1b** (Chart B, dot plot): which regions have the largest millet lean-season rise, and are these the regions where
  the Cadre Harmonisé projects the most people in crisis?

**Audience:** cash-assistance programme officers, who decide when and where to raise transfer amounts.

**Data:** the team's clean tables (run `clean_data.py` first if `clean/` is empty):
`clean/market_month_full.csv` (WFP prices + context, the final dataset) and `clean/ch_region_season.csv`
(Cadre Harmonisé, region × season).

This notebook is the step-by-step version of `a2_seasons.py`: same rules, same numbers, same figures.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display
from matplotlib.colors import LinearSegmentedColormap, Normalize, TwoSlopeNorm
from matplotlib.patches import Rectangle
from scipy.stats import spearmanr

# works whether the notebook is started from a2_seasons/ or from the project folder
ROOT = Path.cwd() if (Path.cwd() / "clean").exists() else Path.cwd().parent
CLEAN, OUT = ROOT / "clean", ROOT / "a2_seasons"

# ---- analysis rules (identical to a2_seasons.py) ----
MIN_MONTHS_PER_YEAR = 10   # a market-year needs >= 10 monthly prices to get a seasonal index
BASE_MONTHS = [1, 2, 3]    # post-harvest baseline: Jan-Mar
LEAN_MONTHS = [7, 8, 9]    # lean-season peak: Jul-Sep
MIN_MONTHS_IN_WINDOW = 2   # a market-year needs >= 2 of the 3 months in each window
MIN_SERIES_YEARS = 10      # a region needs >= 10 millet market-years to be shown in chart B

FOOD_LABEL = {"millet": "Millet", "sorghum": "Sorghum", "maize": "Maize", "cowpeas": "Cowpeas",
              "rice_local": "Local rice", "rice_imported": "Imported rice"}
COUNTRY_ORDER = ["Niger", "Chad", "Burkina Faso", "Mali", "Senegal"]
MONTHS = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

pd.set_option("display.precision", 1)
print("project folder:", ROOT)
```

## 1. Load the prices

Outliers flagged by the cleaning script (12 prices) are excluded. A *series* is one market × one food.

```python
m = pd.read_csv(CLEAN / "market_month_full.csv")
m = m[~m.outlier_flag].copy()
m["series"] = m.market_id.astype(str) + "_" + m.food
print(f"{len(m):,} prices, {m.series.nunique()} series, {m.market_id.nunique()} markets")
m.groupby(["country", "food"]).series.nunique().unstack(fill_value=0)
```

## 2. Seasonal price index

Each price is divided by the average price of **the same market and food in the same calendar year** (× 100).
This removes inflation, price differences between foods and markets, and multi-year shocks (e.g. 2022), so only the
*shape within the year* remains: 100 = that year's average, 107 = 7% above it.
Market-years with fewer than 10 monthly prices are skipped, because a yearly average built from a few months is biased.

```python
g = m.groupby(["series", "year"]).price_lcu
m["months_in_year"] = g.transform("size")
m["year_mean"] = g.transform("mean")
s = m[m.months_in_year >= MIN_MONTHS_PER_YEAR].copy()
s["seasonal_index"] = s.price_lcu / s.year_mean * 100
print(f"{len(s):,} prices with a seasonal index, in {s.groupby(['series', 'year']).ngroups:,} market-years")

heat = s.groupby(["food", "month_of_year"]).seasonal_index.median().unstack()
n_series = s.groupby("food").series.nunique()
amplitude = heat.max(axis=1) - heat.min(axis=1)
heat = heat.loc[amplitude.sort_values(ascending=False).index]
heat.rename(index=FOOD_LABEL, columns=dict(enumerate(MONTHS, 1)))
```

Grains and cowpeas are lowest in January (index 91–92) and highest in August–September (106–108).
Imported rice is flat all year; local rice peaks later (September–October).

### Does the pattern hold in every country?
Same index, millet only, per country (not in the charts, a check).

```python
(s[s.food == "millet"].groupby(["country", "month_of_year"]).seasonal_index.median()
 .unstack().rename(columns=dict(enumerate(MONTHS, 1))))
```

## 3. Lean-season rise per market-year

Rise = mean price in **Jul–Sep** ÷ mean price in **Jan–Mar** of the same year − 1. Both windows need at least 2 of their 3 months.
This is the number a programme officer cares about: how much more the same food costs at the lean-season peak than after harvest.

```python
def window_mean(months):
    w = s[s.month_of_year.isin(months)].groupby(["series", "year"]).price_lcu
    return w.mean().where(w.size() >= MIN_MONTHS_IN_WINDOW)

keys = s.drop_duplicates(["series", "year"]).set_index(["series", "year"])[
    ["food", "country", "region_code", "region_name", "market"]]
rise = keys.assign(rise_pct=(window_mean(LEAN_MONTHS) / window_mean(BASE_MONTHS) - 1) * 100).dropna()
print(f"{len(rise):,} market-years with a lean-season rise")

by_food = rise.groupby("food").rise_pct.describe(percentiles=[.25, .5, .75])[["count", "25%", "50%", "75%"]]
by_food = by_food.rename(index=FOOD_LABEL, columns={"50%": "median %"}).sort_values("median %", ascending=False)
# what the rise means for a fixed cash transfer: price +r% -> the same cash buys 1 - 1/(1+r) less
by_food["same cash buys % less"] = 100 - 100 / (1 + by_food["median %"] / 100)
by_food
```

Note: a price **15% higher** means the same cash buys about **13% less** (1 − 1/1.15), not 15% less.

## 4. Cadre Harmonisé: share of people in phase 3+ by season

Share in phase 3 (crisis) or worse = people in phase 3–5 ÷ population analysed, pooled over the 5 countries
(population-weighted). June–August is only ever a **projection**; Jan–May and Sep–Dec are **current** analyses.
The 13 "current June–August" rows (Burkina Faso's June 2020 update only) are left out so the seasons are comparable.

```python
ch = pd.read_csv(CLEAN / "ch_region_season.csv")
ch = ch[ch.reference_year.between(2015, 2024)]
ch = ch[~((ch.chtype == "current") & (ch.reference_label == "Jun-Aug"))]

def share35(d):
    return d.phase35.sum() / d.population_analysed.sum() * 100

season_share = ch.groupby(["chtype", "reference_label"]).apply(share35, include_groups=False)
print("pooled 2015-2024:")
print(season_share.round(1).to_string())

per_year = pd.DataFrame({
    "Jan-May (current)": ch[(ch.chtype == "current") & (ch.reference_label == "Jan-May")].groupby("reference_year").apply(share35, include_groups=False),
    "Jun-Aug (projected)": ch[ch.chtype == "projected"].groupby("reference_year").apply(share35, include_groups=False),
    "Sep-Dec (current)": ch[(ch.chtype == "current") & (ch.reference_label == "Sep-Dec")].groupby("reference_year").apply(share35, include_groups=False),
})
print("lean-season projection above Jan-May in every year:",
      bool((per_year["Jun-Aug (projected)"] > per_year["Jan-May (current)"]).all()))
per_year
```

### Robustness: only region-years with all three analyses

Some seasons are missing (Mali Jan–May 2022, Niger Jan–May 2024, Burkina Faso Sep–Dec 2024). Because crisis levels rose
over time, missing late seasons could bias the pooled shares. Restricting to region-years where all three analyses exist:

```python
k = ["region_code", "reference_year"]
parts = {"Jan-May (current)": ch[(ch.chtype == "current") & (ch.reference_label == "Jan-May")],
         "Jun-Aug (projected)": ch[ch.chtype == "projected"],
         "Sep-Dec (current)": ch[(ch.chtype == "current") & (ch.reference_label == "Sep-Dec")]}
parts = {n: d.dropna(subset=["phase35"]).set_index(k) for n, d in parts.items()}
common = set.intersection(*(set(d.index) for d in parts.values()))
print(f"{len(common)} region-years with all three analyses")
pd.Series({n: share35(d.loc[sorted(common)]) for n, d in parts.items()}, name="% phase 3+").round(1)
```

## 5. Chart A: heatmap food × month, with the Cadre Harmonisé strip

Colours from the team's palette: a diverging blue–grey–red scale centred on 100 for prices, and a single-hue blue scale for
the Cadre Harmonisé so the two cannot be confused. The figure is saved at 300 dpi to `a2_seasons/fig_a_seasonal_heatmap.png`.

```python
TEXT, TEXT2, SURFACE, GRID = "#0b0b0b", "#52514e", "#fcfcfb", "#e4e3df"
DIVERGING = LinearSegmentedColormap.from_list("div", ["#2a78d6", "#f0efec", "#e34948"])
SEQUENTIAL = LinearSegmentedColormap.from_list(
    "seq", ["#cde2fb", "#9ec5f4", "#6da7ec", "#3987e5", "#256abf", "#184f95", "#0d366b"])
plt.rcParams.update({"font.family": "DejaVu Sans", "font.size": 10, "text.color": TEXT,
                     "axes.labelcolor": TEXT2, "xtick.color": TEXT2, "ytick.color": TEXT2,
                     "axes.edgecolor": GRID, "figure.facecolor": SURFACE, "axes.facecolor": SURFACE,
                     "savefig.facecolor": SURFACE, "figure.dpi": 72})

fig = plt.figure(figsize=(12, 7.2))
gs = fig.add_gridspec(2, 2, height_ratios=[1, 5.2], width_ratios=[40, 1], hspace=0.08, wspace=0.03,
                      left=0.15, right=0.91, top=0.76, bottom=0.12)
ax_ch, ax = fig.add_subplot(gs[0, 0]), fig.add_subplot(gs[1, 0])
cax = fig.add_subplot(gs[1, 1])

norm = TwoSlopeNorm(vmin=88, vcenter=100, vmax=112)
im = ax.imshow(heat.values, cmap=DIVERGING, norm=norm, aspect="auto")
for i in range(heat.shape[0]):
    for j in range(12):
        v = heat.values[i, j]
        ax.text(j, i, f"{v:.0f}", ha="center", va="center", fontsize=9,
                color="white" if abs(v - 100) > 8 else TEXT)
ax.set_xticks(range(12), MONTHS)
ax.set_yticks(range(heat.shape[0]), [f"{FOOD_LABEL[f]}  ({n_series[f]})" for f in heat.index])
ax.tick_params(length=0)
for sp in ax.spines.values():
    sp.set_visible(False)
ax.set_xticks(np.arange(-.5, 12), minor=True); ax.set_yticks(np.arange(-.5, heat.shape[0]), minor=True)
ax.grid(which="minor", color=SURFACE, linewidth=2); ax.tick_params(which="minor", length=0)
ax.add_patch(Rectangle((4.5, -0.5), 4, heat.shape[0], fill=False, edgecolor=TEXT, linewidth=2,
                       linestyle="--", zorder=10, clip_on=False))
ax.set_xlabel("Month   (dashed frame = lean season, June-September)")
ax.set_ylabel("Food  (number of market series)")

cb = fig.colorbar(im, cax=cax)
cb.set_label("Seasonal price index\n(same market & year average = 100)", color=TEXT2)
cb.outline.set_visible(False)

ax_ch.set_xlim(-0.5, 11.5); ax_ch.set_ylim(0, 1); ax_ch.axis("off")
ch_norm = Normalize(0, 12)
for (x0, x1, key, label) in [(-0.5, 4.5, ("current", "Jan-May"), "Jan-May (current)"),
                             (4.5, 7.5, ("projected", "Jun-Aug"), "Jun-Aug (projected)"),
                             (7.5, 11.5, ("current", "Sep-Dec"), "Sep-Dec (current)")]:
    v = season_share[key]
    ax_ch.add_patch(Rectangle((x0 + 0.04, 0.05), x1 - x0 - 0.08, 0.9, color=SEQUENTIAL(ch_norm(v))))
    col = "white" if v > 6 else TEXT
    ax_ch.text((x0 + x1) / 2, 0.63, f"{v:.1f}%", ha="center", va="center", fontsize=12, fontweight="bold", color=col)
    ax_ch.text((x0 + x1) / 2, 0.27, label, ha="center", va="center", fontsize=8.5, color=col)
ax_ch.text(-0.6, 0.5, "Cadre Harmonisé:\n% in phase 3+\n(crisis or worse)", ha="right", va="center",
           fontsize=9, color=TEXT2)

fig.suptitle("Grain prices peak in August-September, in the same season the\nCadre Harmonisé counts the most people in crisis",
             x=0.03, ha="left", fontsize=14, fontweight="bold", y=0.975)
fig.text(0.03, 0.85, "Median seasonal price index per month (each market's price divided by its own average for that year), "
         "Mali, Burkina Faso,\nNiger, Senegal and Chad, 2015-2024. Top strip: share of the population in Cadre Harmonisé "
         "phase 3 (crisis) or worse, pooled 2015-2024.", ha="left", va="center", fontsize=9.5, color=TEXT2)
fig.text(0.03, 0.025, "Sources: WFP food prices (HDX, CC BY-IGO); Cadre Harmonisé (CILSS / HDX, CC BY). Series-years with "
         "fewer than 10 monthly prices and the 12 flagged outliers excluded.\nJune-August is only ever a projection in "
         "the Cadre Harmonisé; it is compared with current estimates for the other months.",
         ha="left", fontsize=8, color=TEXT2)
fig.savefig(OUT / "fig_a_seasonal_heatmap.png", dpi=300)
plt.show()
```

## 6. Lean-season rise per region, and the link to the Cadre Harmonisé

Millet only (the most widely reported staple). Per region: median rise over its market-years and the interquartile range.
Regions with fewer than 10 millet market-years are dropped. The Cadre Harmonisé value per region is the mean projected
June–August share in phase 3+ over 2015–2024.

```python
mil = rise[rise.food == "millet"].reset_index()
reg = (mil.groupby(["country", "region_code", "region_name"]).rise_pct
       .agg(median="median", q25=lambda x: x.quantile(.25), q75=lambda x: x.quantile(.75), n="size")
       .reset_index())
reg = reg.merge(mil.groupby("region_code").series.nunique().rename("n_markets"), on="region_code")
dropped = reg[reg.n < MIN_SERIES_YEARS]
reg = reg[reg.n >= MIN_SERIES_YEARS]
print(f"{len(reg)} regions shown; dropped (< {MIN_SERIES_YEARS} market-years):", ", ".join(dropped.region_name))

proj = ch[(ch.chtype == "projected") & (ch.reference_label == "Jun-Aug")]
reg = reg.merge(proj.groupby("region_code").share_phase35.mean().rename("ch_lean"), on="region_code", how="left")
rho, pval = spearmanr(reg["median"], reg.ch_lean)
print(f"Spearman rank correlation, millet rise vs. Cadre Harmonisé lean-season share: rho = {rho:.2f}, p = {pval:.3f}, n = {len(reg)}")
reg.sort_values("median", ascending=False)
```

ρ = 0.26 is a **weak** positive link and not significant at the 5% level. Regions with the highest crisis shares
(e.g. Sahel in Burkina Faso, Gao in Mali) have only average seasonal rises. A possible explanation, to test in the conflict
question rather than claim here: in conflict areas prices are high or disrupted all year, not only in the lean season.

### Look at the relationship directly (a check, not a submitted chart)

```python
fig, ax = plt.subplots(figsize=(6.5, 4.5))
for c in COUNTRY_ORDER:
    d = reg[reg.country == c]
    ax.scatter(d["median"], d.ch_lean, label=c, s=40)
for _, r in reg[reg.ch_lean > 18].iterrows():
    ax.annotate(r.region_name, (r["median"], r.ch_lean), xytext=(4, 2), textcoords="offset points", fontsize=8)
ax.set_xlabel("Median millet lean-season rise, %")
ax.set_ylabel("CH projected Jun-Aug phase 3+, %")
ax.set_title(f"Regions: seasonal rise vs. crisis share (Spearman rho = {rho:.2f})", loc="left", fontsize=10)
ax.legend(fontsize=8, frameon=False)
plt.show()
```

## 7. Chart B: dot plot with ranges, faceted by country

Saved at 300 dpi to `a2_seasons/fig_b_region_dotplot.png`.

```python
reg["country"] = pd.Categorical(reg.country, COUNTRY_ORDER, ordered=True)
reg = reg.sort_values(["country", "median"], ascending=[True, False])
counts = reg.country.value_counts().reindex(COUNTRY_ORDER)

fig = plt.figure(figsize=(9.5, 14))
gs = fig.add_gridspec(len(COUNTRY_ORDER), 1, height_ratios=counts.values + 1.2, hspace=0.25,
                      left=0.25, right=0.80, top=0.885, bottom=0.07)
ch_max = 25
ch_norm = Normalize(0, ch_max)
axes = []
for k, country in enumerate(COUNTRY_ORDER):
    ax = fig.add_subplot(gs[k], sharex=axes[0] if axes else None)
    axes.append(ax)
    d = reg[reg.country == country].reset_index(drop=True)
    y = np.arange(len(d))
    ax.hlines(y, d.q25, d.q75, color="#bdbcb6", linewidth=2.5, zorder=1)
    ax.scatter(d["median"], y, s=90, c=SEQUENTIAL(ch_norm(d.ch_lean.clip(upper=ch_max))),
               edgecolor=SURFACE, linewidth=1.5, zorder=3)
    ax.set_yticks(y, [f"{r} ({mk} mkts)" for r, mk in zip(d.region_name, d.n_markets)], fontsize=9)
    ax.set_ylim(len(d) - 0.4, -0.6)
    ax.axvline(0, color=TEXT2, linewidth=1)
    med_all = mil[mil.country == country].rise_pct.median()
    ax.axvline(med_all, color=TEXT2, linewidth=1, linestyle=":")
    ax.set_title(f"{country}  -  country median {med_all:.0f}%", loc="left", fontsize=10.5,
                 fontweight="bold", color=TEXT, pad=4)
    ax.grid(axis="x", color=GRID, linewidth=0.8); ax.set_axisbelow(True)
    ax.tick_params(length=0)
    for sp in ["top", "right", "left"]:
        ax.spines[sp].set_visible(False)
    if k < len(COUNTRY_ORDER) - 1:
        ax.tick_params(labelbottom=False)
axes[-1].set_xlabel("Millet price rise, lean season (Jul-Sep) vs. after harvest (Jan-Mar) of the same year, %")
axes[-1].xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:+.0f}%"))
axes[0].set_xlim(-15, 60)

cax = fig.add_axes([0.84, 0.35, 0.022, 0.3])
cb = fig.colorbar(plt.cm.ScalarMappable(norm=ch_norm, cmap=SEQUENTIAL), cax=cax, extend="max")
cb.set_label("Cadre Harmonisé: % of population\nin phase 3+ in June-August\n(projected, mean 2015-2024)", color=TEXT2)
cb.outline.set_visible(False)
cb.ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0f}%"))

fig.suptitle("Millet's lean-season jump is largest in Niger and eastern Chad,\nbut crisis levels follow it only loosely",
             x=0.03, ha="left", fontsize=14, fontweight="bold", y=0.985)
fig.text(0.03, 0.925, f"Dot = median rise across markets and years 2015-2024; bar = middle 50% of market-years. "
         f"Dotted line = country median.\nDot colour = Cadre Harmonisé lean-season share in crisis or worse.\n"
         f"Rank correlation across the {len(reg)} regions: Spearman ρ = {rho:.2f} (p = {pval:.2f}), a weak link.",
         ha="left", va="center", fontsize=9.5, color=TEXT2)
fig.text(0.03, 0.025, "Sources: WFP food prices (HDX, CC BY-IGO); Cadre Harmonisé (CILSS / HDX, CC BY). "
         f"Regions with fewer than {MIN_SERIES_YEARS} millet market-years not shown.\n"
         "A year counts if the market reported at least 10 months, and at least 2 months in each window. Prices are nominal.",
         ha="left", fontsize=8, color=TEXT2)
fig.savefig(OUT / "fig_b_region_dotplot.png", dpi=300)
plt.show()
```

## 8. Findings and limitations

**Findings**
- Millet, sorghum, maize and cowpeas cost about 14–16% more in Jul–Sep than in Jan–Mar of the same year (median), so a fixed
  transfer buys about 13% less food at the lean-season peak. Imported rice is flat; local rice rises less and later.
- The Cadre Harmonisé share in phase 3+ is highest in the June–August projection (7.5% vs. 4.8% Jan–May and 4.5% Sep–Dec,
  pooled), and above the Jan–May share in every year 2015–2024: prices and hunger peak in the same season.
- The seasonal rise is largest in Niger (country median 20%) and eastern Chad, smallest in Senegal (9%).
- Across regions the link between the size of the seasonal rise and the crisis share is weak (Spearman ρ = 0.26, p ≈ 0.07):
  a top-up based on the price calendar alone would miss some of the worst-affected regions.

**Limitations**
- June–August is only ever *projected* in the Cadre Harmonisé; it is compared with *current* estimates for the other months.
  HDX gives the current periods as March–May and October–December, while the file labels them Jan–May and Sep–Dec.
- The yearly normalisation hides price *levels* and shocks on purpose; pooling hides differences between years.
- Regions with 1–2 markets (most of Chad) give less reliable medians; conflict areas report fewer prices.
- Prices are nominal; correlation is not causation.
